```python
path = '/content/drive/MyDrive/P3 T1/Aj.Ton/DATA/confirmed-cases-since-120864.csv'
```

```python
data = pd.read_csv(path)
data.head()
```

```python
this_data = data[['sex','age','province_of_isolation']]
this_data
```

```python
male = this_data[this_data['sex'] == 'ชาย' ]
male['age']
```

```python
female = this_data[this_data['sex'] == 'หญิง' ]
female[['sex','age']]
```

```python
def avg_age(data):
    sum = 0
    N = 0
    for i in data['age']:
        if i > 0:
          sum += i #sum = sum+i
          N += 1
    return sum/N
```

```python
print(f' อายุเฉลี่ยของเพศหญิงคือ {avg_age(female)}')
print(f' อายุเฉลี่ยของเพศชายคือ {avg_age(male)}')
```

```python
kkc = this_data[this_data['province_of_isolation']== 'ขอนแก่น']
kkc
```

```python
kkc_male = this_data[(this_data['province_of_isolation']== 'ขอนแก่น') & (this_data['sex']=='ชาย')]
kkc_male['age']
```

```python
print(f'อายุเฉลี่ยของผู้ป่วยเพศชายในจังหวัดขอนแก่นคือ {avg_age(kkc_male)}')
print(f'อายุเฉลี่ยของผู้ป่วยเพศหญิงในจังหวัดขอนแก่นคือ {avg_age(kkc_female)}')
```
