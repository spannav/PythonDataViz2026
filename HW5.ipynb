HW 5 เขียน function คูณ matrix

แล้ว test กับ matrix ขนาด

- (2,3)*(3,2)
- (4,4)*(4,1)
- (2,2)*(2,2)

```python
import numpy
```

```python
def matric_multi(A,B):
    m = len(A) # จำนวนแถว A
    a = len(A[0]) # จำนวนหลัก A
    b = len(B) #จำนวนแถว B
    n = len(B[0]) # จำนวนหลัก B
    C = numpy.zeros((m,n))
    for i in range(m):
        for j in range(n):
            for sum in range(a):
                C[i,j] = C[i,j] + A[i,sum] * B[sum,j]
    return C
```

```python
print("______matrix ขนาด (2,3)*(3,2)______")
A = numpy.array([[6, 10, 9],[13, 6, 11]])
B = numpy.array([[14, 8],[12, 7],[10, 11]])
matric_multi(A, B)
```

```python
print("______matrix ขนาด (4,4)*(4,1)______")
A = numpy.array([[11, 9, 8, 13],[10, 8, 15, 11],[6, 3, 17, 16],[12, 5, 10, 6]])
B = numpy.array([[8],[14],[2],[11]])
matric_multi(A, B)
```

```python
print("______matrix ขนาด (2,2)*(2,2)______")
A = numpy.array([[17, 9],[2, 13]])
B = numpy.array([[6, 12],[18, 4]])
matric_multi(A, B)
```
