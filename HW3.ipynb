### HW คำนวณเวลาเป็นวินาทีของเวลาต่อไปนี้โดยใช้คำสั่ง split() ช่วย (print ออกมาให้สวยงาม)
12:30:15

13:41:07

12:53:15

00:59:25

11:11:11

16:06:09

```python
t = ['12:30:15','13:41:07','12:53:15','00:59:25','11:11:11','16:06:09']
print(t)
```

```python
h,m,s = t[0].split(':')
total = int(h)*3600 + int(m)*60 + int(s)
print(f'วินาทีของ 12:30:15 คือ {total} sec')
```
